```python
def count_of_vowels(input_text: str) -> int: # нельзя ничего добавить кроме str
    count = 0
    letters = ['a','e','y','u','i','o']
    char_list = list(input_text.lower())
    for ch in char_list:
        if ch in letters:
            count += 1
    print(count)
    return count

count_of_vowels("Kafka")
#count_of_vowels(1) #
```
