Python dictionaries can contain keys and values of different data types

```python
person = {"name": "Eric", (1, 2): ['a', 'b', 'c'], "job": "IT"}
```

Dict API

```python


# print(person[1])  # Raises an error because this key does not exist

# To avoid this, use get()
print(person.get(1))  # Returns None
print(person.get(1, "No"))  # Returns "No"

print("name" in person)  # Check whether this key exists

del person["name"]  # Delete an item

person_keys = person.keys()
print(person_keys)  # Get a view of the dictionary's keys

items = person.items()
print(type(items))  # Get a view of the dictionary's key-value pairs
```

Map iteration

```python
mp = {
    1:1,
    "two":3,
    (3):(1,2,3),
}

for k, v in mp.items(): # iteration by key/value
    print(k,v)

for k in mp.keys(): # iteration by key
    print(k)

for k in mp: # iteration by key
    print(k)

for v in mp.values(): # iteration by value
    print(v)
```

Array merge

```python
dict1 = {"a":1,"c":3}
dict2 = {"a":2,"d":3}
merged = dict1 | dict2 # If u has conflict dict2 перекроет dict1
print(merged)
```
