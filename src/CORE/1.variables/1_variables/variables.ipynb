the data type of variable explicitly.

*Operators*

```python
a = 2; b = 3
print(a + b)  # addition
print(a - b)  # subtraction
print(a * b)  # multiplication
print(a ** b) # exponentiation (power)
print(a // b) # floor division (returns the integer part of the division result)
print(a % b)  # modulo (remainder of division)
```

related operations

```python
print(a / b)   # division (returns float)
print(abs(a))  # absolute value
print(-a)      # negation
a += b  # addition assignment
a -= b  # subtraction assignment
a *= b  # multiplication assignment
a /= b  # division assignment
a //= b # floor division assignment
a %= b  # modulo assignment
a **= b # exponentiation assignment
```
