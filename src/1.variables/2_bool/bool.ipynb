**Логические операторы**

bool может иметь только 2 значения - True / False

```python
t = True
f = False
print(t)
print(f)

isTrue = 3 == 4
print(isTrue)
```

Можно проверить массив/переменную и тд если пустая или значение по умолчанию = F

```python
string_bool_1 = ''
print(bool(string_bool_1)) # Пустой объект = False
```

*Работа с f строками*

```python
warm = True
sunny = False

print(f'Тепло {warm}, Солнечно {sunny}')
going_to_the_beach = warm and sunny
print(f"Едем на пляж ? {going_to_the_beach}")
```

*Приоритет операторов*

not - самый высокий приоритет

and - ниже

or - самый низкий приоритет

```python
res = (False or True) and not (False and True or True)
print(res)
```
