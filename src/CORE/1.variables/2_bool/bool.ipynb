**Boolean operators**

Boolean can have only two values: True or False

```python
t = True
f = False
print(t)
print(f)

isTrue = 3 == 4
print(isTrue)
```

Check whether an array or variable is empty or has the default value

```python
string_bool_1 = ''
print(bool(string_bool_1)) # empty object = False
```

*Working with f strings*

```python
warm = True
sunny = False

print(f'Warm {warm}, Sunny {sunny}')
going_to_the_beach = warm and sunny
print(f"Go to the beach ? {going_to_the_beach}")
```

*Operators priority*

not - The highest priority

and - lower

```python
res = (False or True) and not (False and True or True)
print(res)
```
