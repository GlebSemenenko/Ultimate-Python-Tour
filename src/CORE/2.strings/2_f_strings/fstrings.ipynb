*F Strings*

```python
name = "Alice"
age = 25
height = 1.75

# Basic formatting
print(f"My name is {name}, I am {age} years old")
print(f"My height: {height} m")

# Expressions inside f-strings
print(f"In 5 years I will be {age + 5} years old")
print(f"My name in uppercase: {name.upper()}")
print(f"Square of number: {5**2}")
```

Number Formatting

```python
price = 1234.5678
negative_price = -1234.5678

# Rounding
print(f"Price: {price:.2f}")      # "1234.57" (2 decimal places)
print(f"Price: {price:.0f}")      # "1235" (integer)
print(f"Price: {price:.3f}")      # "1234.568"

# Alignment
print(f"Price: {price:10.2f}")    # "   1234.57" (total 10 characters)
print(f"Price: {price:<10.2f}")   # "1234.57   " (left alignment)
print(f"Price: {price:>10.2f}")   # "   1234.57" (right alignment)
print(f"Price: {price:^10.2f}")   # " 1234.57  " (center alignment)

# Number signs
print(f"Positive: {price:+}")      # "+1234.5678"
print(f"Negative: {negative_price:+}")  # "-1234.5678"
print(f"With space: {price: }")    # " 1234.5678" (space for positive)
print(f"Minus only: {negative_price:-}")  # "-1234.5678" (sign only for negative)

# Thousands separators
print(f"With separator: {price:,.2f}")   # "1,234.57"
print(f"With separator: {price:_.2f}")   # "1_234.57"
```

Date and Time Formatting

```python
from datetime import datetime

now = datetime.now()
print(f"Current time: {now}")
print(f"Date: {now:%Y-%m-%d}")        # "2024-01-15"
print(f"Time: {now:%H:%M:%S}")        # "14:30:45"
print(f"Format: {now:%d.%m.%Y %H:%M}") # "15.01.2024 14:30"
```
