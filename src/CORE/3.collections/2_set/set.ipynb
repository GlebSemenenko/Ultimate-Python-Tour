*Set*

"A set contains unique elements. Order is not guaranteed. Only immutable values can be stored."

"This is a set of unique elements. The storage order is not guaranteed. It can only hold immutable values."

"A set stores only unique elements. The order is not preserved. Only immutable objects can be added."

```python
# Set Creating
# You can create a set from any iterable object

empty_set = set()
set_from_values = {1,2,3,3}
print(set_from_values)

arr = [1,2,3,3,3,4]
l_set = set(arr) # from list
print(l_set)

st = "aabbcc"
str_set = set(st) # из строки
print(str_set)
```
