Adding element

```python
my_list = [1, 2, 3, 4]       # Create initial list with elements 1, 2, 3, 4
my_list.insert(0, 99)        # Insert at index 0, now list: [99, 1, 2, 3, 4]
my_list.extend([5, 6])       # Add elements 5 and 6 from another list, now list: [99, 1, 2, 3, 4, 5, 6]
```

Removing elements

```python
my_list.pop()                # Remove and return the last element, now list: [99, 1, 2, 3, 4, 5]
my_list.pop(1)               # Remove and return element at index 1, now list: [99, 2, 3, 4, 5]
del my_list[0]               # Delete element at index 0, now list: [2, 3, 4, 5]
my_list.remove(3)            # Remove the first element with value 3 (REMOVAL BY VALUE), now list: [2, 4, 5]
#my_list.clear()             # Remove all elems
```

Sorting and ordering

```python
my_list.sort(reverse=True)   # Sort in descending order, now list: [5, 4, 3, 2, 1, 0]
my_list.sort()               # Sort elements in-place (ascending), now list: [0, 1, 2, 4, 5]
my_list.copy()               # Create a shallow copy of the list, new_list: [0, 1, 2, 4, 5]
my_list.reverse()            # Reverse the list in-place
```

List Compare

```python
a = [1,2,3]
b = [1,5,6]
# 1. Simple equality (order + values)
res = a == b  # True if identical elements in same order

# 2. Ignore order (sort or use sets)
res = sorted(a) == sorted(b)  # or set(a) == set(b) (loses duplicates)

# 3. Find differences (what's in A, but not in B)
res = [x for x in a if x not in b]  # elements only in first list

# 4. Element-wise comparison (position by position)
resn = [x == y for x, y in zip(a, b)]  # returns boolean mask

# 6. Nested structures (lists of dicts, etc.)
a == b  # works recursively for any depth
```
