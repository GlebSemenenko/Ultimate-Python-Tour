*Strings*

Creating Strings

```python
# Basic ways to create strings
string1 = 'Single quotes'
string2 = "Double quotes"
string3 = """Multi-line
text with
line breaks"""
string4 = '''Also multi-line
text'''

# Important: Python doesn't have a separate char type, a character is a string of length 1
char = 'A'
print(type(char))  # <class 'str'>
```

Escape Sequences

```python
# Basic escape sequences
print("String with\nline break")        # \n - newline
print("String with\ttab")               # \t - tab
print("File path: C:\\Users\\Name")     # \\ - backslash
print('Single quote: \'')               # \' - single quote
print("Double quote: \"")               # \" - double quote

# Raw strings (ignore escape sequences)
raw_string = r"C:\Users\Name\Documents"
print(raw_string)  # C:\Users\Name\Documents
```

Type Conversion

```python
# Converting number to string
number = 123
number_str = str(number)
print(f"Type: {type(number_str)}")  # <class 'str'>
print(f"String length: {len(number_str)}")  # 3
```

Basic String Methods

```python
txt = """
How many chances in life you weigh, \nHow many words you save,
The perfect place is "here",
The perfect moment is "now".
"""

# Checking for substring
print("moment" in txt)  # True

# Concatenation (string addition)
print(txt + txt)  # Duplicates the text

txt = "  Hello, World!  "

# Case conversion
print(txt.upper())      # "  HELLO, WORLD!  "
print(txt.lower())      # "  hello, world!  "
print(txt.capitalize()) # "  hello, world!  "
print(txt.title())      # "  Hello, World!  "

# Whitespace removal
print(txt.strip())      # Removes whitespace from both sides
print(txt.lstrip())     # Removes whitespace from left
print(txt.rstrip())     # Removes whitespace from right

# Substring replacement
print(txt.replace("World", "Python"))  # "  Hello, Python!  "

# Search and count
print(txt.count("l"))    # Count of "l" characters (3)
print(txt.find("World")) # Find substring (returns index)
print(txt.index("World"))# Same but raises error if not found

# Split and join
words = txt.split()      # Splits by whitespace
print(words)             # ['Hello,', 'World!']
print("-".join(words))   # "Hello,-World!"

# Start and end checks
print(txt.startswith("  Hello"))  # True
print(txt.endswith("!  "))        # True
```

Slicing

```python
text = "Python Programming"

# Basic slices
print(text[0])      # First character "P"
print(text[-1])     # Last character "g"
print(text[0:6])    # "Python" (0 to 5)
print(text[7:])     # "Programming" (7 to end)
print(text[:6])     # "Python" (start to 5)
print(text[7:18])   # "Programming" (7 to 17)

# Slices with step
text2 = "aabbccddffgghhjj"
print(text2[::2])   # "abcdgfhj" (every 2nd character)
print(text2[1::2])  # "abcdfghj" (starting from index 1)
print(text2[::-1])  # "jjhhggffddccbbaa" (reverse string)
```
