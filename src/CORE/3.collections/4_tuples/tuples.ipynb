Tuple API

```python
tupl = ('a','b','c')
# tupl.insert(1,1) # u cant insert or change value

l = [1,2,2,5]
l_tuple = tuple(l) # tuple from another collection

single_item = (42,)  # tuple with one element (, - must have !)

el = tupl[0] # use tuple element
```

```python
t = ([1, 2, 3], 1)

t[0].append(4)
t[0][0] = 3  # Access the list nested inside the tuple

print(t[0][0])
print(t)

# If a tuple contains a list, the list can be modified
tuple_with_list = ("a", "b", [1, 2, 3])

tuple_with_list[2].append(4)
print(tuple_with_list[2])  # The list has been modified

a = tuple_with_list[2]
a[2] = 1000000

# This is how you can modify the elements of a list inside a tuple
print(a, tuple_with_list)
```
